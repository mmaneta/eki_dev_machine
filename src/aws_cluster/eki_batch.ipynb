```python
from aws_cluster import eki_batch
```

## Read Batch Job configuration file

The configuration file requires that the compute infrastructure, task definition, and job queue for the Batch Job are already configured

```python
fn_yaml = ('s3://scratch-marco/test_batch_utils.yaml')
```

## Create the Batch Job and configure environment

Run the cell below only once. It will create a NAT gateway in EKI's public subnet and allocate a public elastic IP. This is necessary to allow the jobs launched in the private subnets can register with the cluster

```python

## Do not modify below this line
ekibatch = eki_batch.EkiBatch(fn_yaml)
ekibatch.create_nat_gateway() 
```

# Inspect the jobs that will be sent to the queue

The cell below will display the number of jobs that will be sent to the job queue and the command line that will be run to launch each of the jobs. 

The results of the cell will display the instructions as they will be sent to the container running the job. Make sure they are correct before actually sending the jobs 

```python
## Do not modify below this line
ekibatch.print_command_list()
```

## Run the jobs

```python
run_sample_job = 1

## Do not modify below this line
if ekibatch.nat_gateway is not None:
    jobs = ekibatch.submit_jobs(run_sample_job)
    print(f"Job {jobs['jobName']} with id {jobs['jobId']} submitted")
```
